```python
import os
import pandas as pd
import requests
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from dotenv import load_dotenv
```

```python
load_dotenv()
base_url = "https://api.insee.fr/api-sirene/3.11/siret/"
```

```python
sirene_api_key = os.environ.get("SIRENE_API_KEY")
```

```python
headers = {
    "X-INSEE-Api-Key-Integration": sirene_api_key
}
# Parameters for the query
params = {
    "q": "activitePrincipaleUniteLegale:86.21Z AND trancheEffectifsUniteLegale:[0 TO 53]",
    "champs": "siret,nombrePeriodesEtablissement,activitePrincipaleUniteLegale,trancheEffectifsUniteLegale"
}
response = requests.get(
    url=base_url,
    headers=headers,
    params=params
)
```
